```python
import pandas as pd

url = "https://raw.githubusercontent.com/GehadGad/Spam-and-ham-data/master/Data.csv"

df = pd.read_csv(url)

df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
import nltk
from nltk import tokenize
from nltk.tokenize import RegexpTokenizer
from sklearn.model_selection import train_test_split
```

```python
# Faz o download da lista de stopwords
nltk.download('stopwords')

# Tokenizador que separa o texto por espaços em branco
token_espaco = tokenize.WhitespaceTokenizer()

# Tokenizador que mantém apenas caracteres alfanuméricos (remove pontuação)
token_espaco = RegexpTokenizer(r'\w+')

# Carrega a lista de stopwords em inglês (palavras irrelevantes)
palavras_irrelevantes = nltk.corpus.stopwords.words("english")
```

```python
# Lista para armazenar as frases após o pré-processamento
frases_processadas = list()

# Percorre cada registro do dataset
for mensagem in df.Message:

  # Lista temporária para armazenar palavras relevantes da frase
  nova_frase = list()

  # Tokeniza o texto em palavras
  palavras_texto = token_espaco.tokenize(mensagem)

  # Filtra as palavras removendo as stopwords
  for palavra in palavras_texto:
    if palavra not in palavras_irrelevantes:
      nova_frase.append(palavra)

  # Reconstrói a frase sem as palavras irrelevantes
  frases_processadas.append(' '.join(nova_frase))

# Adiciona a coluna com o texto processado ao DataFrame
df['tratamento_1'] = frases_processadas
```

```python
df.head()
```

 # Treino Naive Bayes


```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
naive_bayes = MultinomialNB()
vetorizar = CountVectorizer(lowercase=False)

bag_of_words = vetorizar.fit_transform(df.tratamento_1)

tfidf = TfidfVectorizer(lowercase=False)
vetor_tfidf = tfidf.fit_transform(df['tratamento_1'])

treino, teste, classe_treino, classe_teste = train_test_split(
    bag_of_words,  #trocar por bag_of_words (opcional)
    df.Category,
    test_size = 0.3,
    random_state = 42
)

naive_bayes.fit(treino, classe_treino)
```

```python
y_test_predictions = naive_bayes.predict(teste)

precision = precision_score(
    classe_teste,
    y_test_predictions,
    pos_label='spam'
)

recall = recall_score(
    classe_teste,
    y_test_predictions,
    pos_label='spam'
)

f1score = f1_score(
    classe_teste,
    y_test_predictions,
    pos_label='spam'
)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1score)
```

# Treino Regressão Logistica

```python
from sklearn.linear_model import LogisticRegression

regressao_logistica = LogisticRegression() # Cria o modelo de Regressão Logística
vetorizar = CountVectorizer(lowercase=False) # Cria o vetorizador Bag-of-Words sem converter para minúsculas

bag_of_words = vetorizar.fit_transform(df.tratamento_1) # Converte os textos processados em matriz de contagem de palavras

tfidf = TfidfVectorizer(lowercase=False) # Cria o vetorizador TF-IDF sem converter para minúsculas
vetor_tfidf = tfidf.fit_transform(df["tratamento_1"]) # Converte os textos processados em representação TF-IDF

treino, teste, classe_treino, classe_teste = train_test_split(
    bag_of_words,  #trocar por bag_of_words (opcional)
    df.Category,
    test_size=0.3,
    random_state=42
)

regressao_logistica.fit(treino, classe_treino) # Treina o modelo de Regressão Logística com os dados de treino
```

```python
y_test_predictions = regressao_logistica.predict(teste)

precision = precision_score(
    classe_teste,
    y_test_predictions,
    pos_label='spam'
)

recall = recall_score(
    classe_teste,
    y_test_predictions,
    pos_label='spam'
)

f1score = f1_score(
    classe_teste,
    y_test_predictions,
    pos_label='spam'
)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1score)
```
