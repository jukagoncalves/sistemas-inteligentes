# Descrição do dataset

```python
"""
Wine Dataset - Explicação

O Wine Dataset é um conjunto de dados clássico de aprendizado de máquina, frequentemente utilizado em problemas
de classificação. Ele contém informações químicas sobre amostras de vinho produzidas a partir de três diferentes
cultivares de uvas na mesma região da Itália.

Estrutura do Dataset:
- Observações: 178 amostras de vinho.
- Características (Features): 13 atributos contínuos que descrevem propriedades químicas e físicas do vinho.
- Classes (Target): 3 categorias, representando os cultivares de uvas:
  1. Classe 0: Vinho produzido pelo cultivar 1.
  2. Classe 1: Vinho produzido pelo cultivar 2.
  3. Classe 2: Vinho produzido pelo cultivar 3.

Descrição das Características:
1. Alcohol: Teor alcoólico do vinho (%).
2. Malic acid: Concentração de ácido málico (g/l), relacionado à acidez do vinho.
3. Ash: Quantidade de cinzas (g/l), uma medida dos minerais presentes.
4. Alcalinity of ash: Alcalinidade das cinzas, indicando pH relativo.
5. Magnesium: Concentração de magnésio (mg/l), importante para o sabor e estabilidade.
6. Total phenols: Teor total de fenóis, compostos antioxidantes que afetam cor e sabor.
7. Flavanoids: Subgrupo de fenóis, associado à intensidade e estrutura do sabor.
8. Nonflavanoid phenols: Fenóis não flavonoides, que contribuem para a acidez.
9. Proanthocyanins: Tipo de composto fenólico ligado à adstringência.
10. Color intensity: Intensidade da cor, avaliada visualmente.
11. Hue: Matiz do vinho, uma relação entre intensidade e tom da cor.
12. OD280/OD315 of diluted wines: Proporção da absorção óptica em comprimentos de onda diferentes, relacionada à
    concentração de compostos fenólicos.
13. Proline: Concentração de prolina (mg/l), um aminoácido que pode indicar maturidade da uva.

Classes (Rótulos):
- Classe 0 (Cultivar 1): Representa vinhos de um tipo específico de uva. Geralmente mais ricos em álcool e fenóis.
- Classe 1 (Cultivar 2): Vinhos que podem apresentar maior concentração de ácido málico e menos fenóis.
- Classe 2 (Cultivar 3): Vinhos que frequentemente têm maior intensidade de cor e prolina.

Aplicação no Machine Learning:
- Propósito:
  Classificar vinhos com base nas suas características químicas.
  Identificar padrões químicos para diferenciar os cultivares.
- Tarefas comuns:
  Problemas de classificação supervisionada.
  Análise de importância das características para prever a classe.

Este dataset é amplamente usado para explorar métodos de classificação e análise de dados.
"""
```

# Carrergando bibliotecas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, ConfusionMatrixDisplay
```

# Carrergando Dataset

```python
# Carregar o dataset Wine
# O dataset contém informações químicas de diferentes vinhos classificados em 3 categorias
wine = load_wine()
X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = pd.Series(wine.target, name="wine_class")
```

# Dividir os dados em treino e teste (o nome dessa divisão é houldout)

```python
x_train, x_test, y_train, y_test = train_test_split( X,
                                                      y,
                                                     test_size = 0.3,
                                                     shuffle = True)
```

# Normalizar os dados

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_test
```

# Escolher o melhor K

```python
k_values = range(1,21) #testar para k entre 1 e 10
accuracies = []
precisions = []
recalls = []

#encontrar metricas para diferentes valores de x
for k in k_values:
  knn = KNeighborsClassifier(n_neighbors= k)
  knn.fit(x_train, y_train)
  y_pred = knn.predict(x_test)

  accuracies.append(accuracy_score(y_test, y_pred))
  precisions.append(precision_score(y_test, y_pred, average='weighted'))
  recalls.append(recall_score(y_test, y_pred, average= 'weighted'))

  # Plotar as métricas em função de K
plt.figure(figsize=(10, 6))
plt.plot(k_values, accuracies, marker='o', label="Acurácia")
plt.plot(k_values, precisions, marker='s', label="Precision")
plt.plot(k_values, recalls, marker='^', label="Recall")
plt.title("Métricas para diferentes valores de K no KNN", fontsize=14)
plt.xlabel("Número de Vizinhos (K)", fontsize=12)
plt.ylabel("Valor da métrica", fontsize=12)
plt.xticks(k_values)
plt.legend()
plt.grid()
plt.show()
```

# Treinar Algoritmo com Melhor K

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train, y_train)
```

# Classificar os dados de teste e exibir a acurácia

```python
#faz previsao atraves das caracteristicas de teste
y_pred = knn.predict(x_test)
print(y_pred)
print(y_test.values)
```

```python
print(f'Percentual de acurácia: {accuracy_score(y_test, y_pred)*100}%')
print(f'Percentual de precision: {precision_score(y_test, y_pred, average='weighted')}%')
print(f'Percentual de recall: {recall_score(y_test, y_pred, average="weighted")}%')
```
